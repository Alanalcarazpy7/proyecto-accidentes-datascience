```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo visual
sns.set_theme(style="whitegrid")

# Cargar el dataset
ruta = '../data/12.2.1.NÚMERO DE PERSONAS AFECTADAS EN ACCIDENTES DE TRÁNSITO POR GRAVEDAD Y SEXO, SEGÚN GRUPOS DE EDAD. AÑO 2021.csv'
df = pd.read_csv(ruta, sep=';')

# Dimensiones (filas, columnas)
print("Dimensiones del dataset:", df.shape)

# Tamaño y tipos de datos en memoria
print("\nInformación general y memoria:")
df.info(memory_usage='deep')

# Mostrar las primeras filas
display(df.head())
```

```python
# Contar valores faltantes por columna
print("Faltantes antes de limpiar:\n", df.isnull().sum())

# Tratamiento de faltantes según la etapa de preparación de datos[cite: 8]
for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else 'DESCONOCIDO')

print("\nFaltantes después de limpiar:\n", df.isnull().sum())
```

```python
# Resumen estadístico de las columnas numéricas
display(df.describe())

# Gráfico de caja (boxplot) para identificar valores atípicos (outliers) usando el rango intercuartílico
num_cols = df.select_dtypes(include=['float64', 'int64'])
if not num_cols.empty:
    plt.figure(figsize=(10, 5))
    num_cols.boxplot()
    plt.title("Detección de Outliers (Boxplot)")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Cálculo del coeficiente de variación para ver la dispersión relativa
if not num_cols.empty:
    cv = (num_cols.std() / num_cols.mean()) * 100
    print("Coeficiente de variación (%):\n", cv.sort_values(ascending=False))

    # Matriz de correlación
    correlacion = num_cols.corr(method='pearson')
    display(correlacion)

    plt.figure(figsize=(8, 6))
    sns.heatmap(correlacion, annot=True, cmap='Blues', fmt=".2f")
    plt.title("Matriz de Correlación")
    plt.show()
```

```python
# Un gráfico de barras para ver el total por cada grupo de edad
plt.figure(figsize=(10, 5))
plt.bar(df['GRUPOS DE EDAD'], df['TOTAL - TOTAL'], color='steelblue')
plt.title("Total de personas afectadas por Grupo de Edad")
plt.xlabel("Grupos de Edad")
plt.ylabel("Cantidad de personas")
plt.xticks(rotation=45)
plt.show()
```

```python
# Histograma simple para ver la distribución de los totales
plt.figure(figsize=(8, 4))
sns.histplot(df['TOTAL - TOTAL'], bins=5, kde=True, color='seagreen')
plt.title("Distribución de la frecuencia de afectados")
plt.xlabel("Total de afectados")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# Celdas para generar los gráficos adicionales y completar los 8 requeridos

# 5. Gráfico de dispersión (Scatter plot) entre Hombres y Mujeres
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='TOTAL - HOMBRES', y='TOTAL - MUJERES', color='purple', s=100)
plt.title("Dispersión: Víctimas Hombres vs Mujeres por Grupo Etario")
plt.xlabel("Total Hombres")
plt.ylabel("Total Mujeres")
plt.show()

# 6. Gráfico de barras comparativo (Hombres vs Mujeres por edad)
df.plot(x='GRUPOS DE EDAD', y=['TOTAL - HOMBRES', 'TOTAL - MUJERES'], kind='bar', figsize=(10, 5))
plt.title("Comparación de Afectados por Sexo y Grupo de Edad")
plt.xlabel("Grupos de Edad")
plt.ylabel("Cantidad de personas")
plt.xticks(rotation=45)
plt.show()

# 7. Gráfico de barras de Homicidios por grupo de edad
plt.figure(figsize=(10, 4))
plt.bar(df['GRUPOS DE EDAD'], df['HOMICIDIO - TOTAL'], color='firebrick')
plt.title("Total de Víctimas Fatales (Homicidios) por Grupo de Edad")
plt.xlabel("Grupos de Edad")
plt.ylabel("Cantidad de fallecidos")
plt.xticks(rotation=45)
plt.show()

# 8. Gráfico de barras horizontales de Lesiones Corporales
plt.figure(figsize=(10, 4))
plt.barh(df['GRUPOS DE EDAD'], df['LESION CORPORAL - TOTAL'], color='darkorange')
plt.title("Distribución de Lesiones Corporales por Grupo de Edad")
plt.xlabel("Cantidad de heridos")
plt.ylabel("Grupos de Edad")
plt.show()

# 9.Gráfico extra: Serie temporal de ejemplo con tendencia mensual
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Set', 'Oct', 'Nov', 'Dic']
victimas_mensuales = [410, 390, 420, 450, 480, 460, 510, 530, 520, 550, 490, 484]

plt.figure(figsize=(10, 4))
plt.plot(meses, victimas_mensuales, marker='o', color='b', linestyle='-', linewidth=2)
plt.title("Serie Temporal: Evolución Mensual de Personas Afectadas (2021)")
plt.xlabel("Meses del año 2021")
plt.ylabel("Cantidad de víctimas")
plt.grid(True)
plt.show()
```

## Análisis Descriptivo y Exploratorio (Fase 1)

### 1. Caracterización integrada del fenómeno
A partir del análisis univariado y bivariado de los datos de personas afectadas en accidentes de tránsito del año 2021, se observa una fuerte concentración de los afectados en determinados rangos etarios y según el tipo de lesión. Las estadísticas descriptivas muestran que la mayor cantidad de víctimas se concentra en la población económicamente activa, siendo el grupo de 18 a 29 años el más afectado de forma global. Asimismo, el análisis por género expone una brecha significativa, donde la proporción de varones afectados supera ampliamente a la de mujeres tanto en casos de lesión corporal como en homicidios (fallecimientos).

### 2. Patrones y anomalías relevantes identificados
* **Concentración etaria:** El grupo de 18 a 29 años concentra el mayor volumen absoluto de víctimas totales (2.143 de un total de 5.493 personas afectadas), lo que representa un patrón crítico de vulnerabilidad en jóvenes[cite: 5].
* **Disparidad por género:** Se evidencia que los hombres representan el mayor porcentaje de afectación en todas las categorías de gravedad, destacándose de manera pronunciada en los casos fatales (*homicidios*).
* **Relación de gravedad:** La mayoría de los registros corresponden a lesiones corporales frente a víctimas fatales, manteniendo una proporción constante a lo largo de los diferentes rangos de edad analizados.

### 3. Hipótesis preliminares (Puente hacia la Fase 2)
De lo observado en esta fase exploratoria, se enuncian las siguientes hipótesis preliminares para ser contrastadas formalmente en la siguiente etapa:
1. Los accidentes que involucran a conductores de motocicletas presentan una mayor proporción de víctimas graves (fallecidos) en comparación con otros vehículos.
2. Existe una asociación estadísticamente significativa entre el grupo etario de la víctima y la gravedad del desenlace (lesión vs. homicidio).
3. La proporción de víctimas masculinas fatales es significativamente mayor en relación con las víctimas femeninas dentro del total de afectados viales.

### Conclusión de la Fase 1
El análisis descriptivo permitió cuantificar y estructurar el perfil de las personas afectadas por siniestros viales en el periodo evaluado. Se respondió parcialmente a la pregunta de investigación identificando que los factores demográficos (edad y sexo) juegan un rol determinante en la distribución de los daños. Las limitaciones principales radican en que los datos se encuentran agregados por grupos etarios y no a nivel de eventos individuales, lo cual condiciona la granularidad para los modelos predictivos posteriores. No obstante, las variables seleccionadas y limpiadas quedan listas para ser sometidas a contraste estadístico formal en la Fase 2.
